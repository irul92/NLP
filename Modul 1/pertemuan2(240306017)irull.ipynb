```python
# import library
import pandas as pd
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
#load data
df = pd.read_csv("IndoLeX_Database.csv")
text_column = "category"
#memecah teks jadi kalimat
text = "Saya belajar NLP. NLP sangat menarik. Saya ingin menjadi NLP Engineer."
sentences = sent_tokenize(text)
for sentence in sentences:
    print(sentence)
#Word Tokenization
from nltk.tokenize import word_tokenize
text = "Saya belajar Natural Language Processing."
tokens = word_tokenize(text)
print(tokens)
```

```python
#Character Tokenization
text = "NLP"
characters = list(text)
print(characters)

#Menerapkan word tokenization ke seluruh data:
df["tokens"] = df[text_column].fillna("").astype(str).apply(word_tokenize)
df[[text_column, "tokens"]].head()
df["jumlah_token"] = df["tokens"].apply(len)
df["jumlah_token"].describe()

#membandingkan jumlah kata dan token
df["jumlah_kata"] = df[text_column].astype(str).apply(
    lambda x: len(x.split())
)
df[["jumlah_kata", "jumlah_token"]].describe()
```

```python
import spacy

# Load model spaCy
nlp = spacy.load("xx_ent_wiki_sm")

# Teks yang akan dibandingkan
texts = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000"
]

# Membuat data tabel
data = []

for text in texts:
    # NLTK
    nltk_tokens = word_tokenize(text)

    # split()
    split_tokens = text.split()

    data.append([
        text,
        nltk_tokens,
        split_tokens,
    ])

# Membuat tabel
tabel = pd.DataFrame(
    data,
    columns=[
        "Teks",
        "NLTK word_tokenize",
        "split()",
    ]
)

# Menampilkan tabel
display(tabel)
```
